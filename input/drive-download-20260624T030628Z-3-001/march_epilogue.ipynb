```python
!pip install cloudscraper
import cloudscraper
import urllib3
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from time import sleep
from time import time
import requests
from google.colab import files
from bs4 import Comment
from bs4.element import Tag
# header: str = "https://www.basketball-reference.com"

def soupify(url: str):

    http = urllib3.PoolManager()
    response = http.request('GET', url)
    soup = BeautifulSoup(response.data)

    return soup

def comment_helper(soup, obj: str):

    for comment in soup.find_all(string=lambda string: isinstance(string, Comment)):
        if comment.find(f"<{obj}") > 0: # and comment.find("totals_stats_sh") > 0:
            comment_soup = BeautifulSoup(comment, 'html.parser')
            res = comment_soup.find(obj)

    return res

def comment_helper(soup, obj: str, id = None):

    for comment in soup.find_all(string=lambda string: isinstance(string, Comment)):
        if not id:
            if comment.find(f"<{obj}") > 0:
                comment_soup = BeautifulSoup(comment, 'html.parser')
                res = comment_soup.find(obj)
        else:
            if comment.find(f"<{obj}") > 0 and comment.find(id) > 0:
                comment_soup = BeautifulSoup(comment, 'html.parser')
                res = comment_soup.find(obj)

    return res

def soupify_with_cloudscraper(url):
    scraper = cloudscraper.create_scraper()
    response = scraper.get(url)
    return BeautifulSoup(response.text, 'html.parser')

def get_table_ids(soup: "BeautifulSoup") -> list[str]:

    return list(t.get('id') for t in soup.find_all('table') if t.get('id'))

def get_columns(table_soup: "BeautifulSoup") -> list[str]:

    return list(t.get('data-stat') for t in table_soup.find('thead').find_all('th'))
```

```python
def get_tourney_teams(bracket_page, year, header = "https://sports-reference.com"):

    nec = list(d.find_all('a') for d in bracket_page.find('div', {"id" : "brackets"}).find_all('div') if not d.find('p'))
    all_teams = list(item for L in nec for item in L if "schools" in item["href"])
    apps = {team.contents[0]:header + team["href"].replace(f"{year}", f"{year}-gamelogs") for team in all_teams}

    return apps
```

```python
first_year = 2010
start = time()
for year in range(2010, 2011):
    url: str = f"https://www.sports-reference.com/cbb/postseason/men/{year}-ncaa.html"
    s = soupify_with_cloudscraper(url)
    sleep(5)
    urls = get_tourney_teams(s, year)
    for team in list(urls):
        log = soupify_with_cloudscraper(urls[team])
        t = log.find('table', {"id" : "team_game_log"})
        # if team == next(iter(urls)):
        if year == first_year and team == next(iter(urls)):
            cols = list(c for c in get_columns(t)[14:] if "pct" not in c)
            df = pd.DataFrame(columns = ["year", "team","G", "MP", "pts", "opp_pts"] + cols)
        tracker = {c:0 for c in cols}
        pts = 0
        opp_pts = 0
        mins = 0
        games = 0
        for game in t.find('tbody').find_all('tr'):
            if game.find('td'):
                if game.find('td', {"data-stat" : "opp_name_abbr"}).find('a'):
                    if any(y in game.find('td', {"data-stat" : "game_type"}).contents[0] for y in {"REG", "CTOURN"}):
                        if game.find('td', {"data-stat" : "opp_fg"}).contents:
                            tracker = {col: float(game.find('td', {"data-stat" : col}).contents[0]) + tracker[col] for col in cols}
                            pts += int(game.find('td', {"data-stat" : "team_game_score"}).contents[0])
                            opp_pts += int(game.find('td', {'data-stat' : "opp_team_game_score"}).contents[0])
                            games += 1
                            mins += 40
                            if game.find('td', {"data-stat" : "overtimes"}).contents:
                                OT = game.find('td', {"data-stat" : "overtimes"}).contents[0]
                                if OT[0].isnumeric():
                                    mins += int(OT[0])*5
                                else:
                                    mins += 5
                        else:
                            print([year, team,
                                  # game.find('td', {"data-stat" : "team_game_num_season"}).contents[0]])
                                  game.find('td', {"data-stat" : "opp_name_abbr"}).find('a').contents[0],
                                  game.find('td', {"data-stat" : "date"}).contents[0]
                            ])
                    else:
                        break
        df.loc[len(df.index)] = [year, team, games, mins, pts, opp_pts] + list(tracker.values())
        sleep(5)
    print(f"{year} done, time elapsed: {time() - start}")
```
